```python
import pandas as pd
```

```python
dt_docente = pd.read_csv(
    "dados/Tabela_Docente_2025_V2.csv",
    sep=";",
    encoding="latin1")
```

```python
doc_agg = dt_docente.groupby(['NO_UF','TP_LOCALIZACAO','TP_DEPENDENCIA'], as_index=False).agg(
    total_doc=('QT_DOC_BAS','sum'),
    total_doc_superior=('QT_DOC_BAS_ESCO_SUP_GRAD','sum'),
    total_doc_concursado=('QT_DOC_BAS_VINCULO_CONCUR','sum'),
)
doc_agg['pct_doc_superior'] = doc_agg['total_doc_superior'] / doc_agg['total_doc']
```

```python
dt_escola = pd.read_csv(
    "dados/Tabela_Escola_2025_V2.csv",
    sep=";",
    encoding="latin1"
)
```

```python
escola_agg = dt_escola.groupby(['NO_UF','TP_LOCALIZACAO','TP_DEPENDENCIA'], as_index=False).agg(
    pct_biblioteca=('IN_BIBLIOTECA','mean'),
    pct_internet_alunos=('IN_INTERNET_ALUNOS','mean'),
    pct_lab_ciencias=('IN_LABORATORIO_CIENCIAS','mean'),
    pct_acessibilidade_rampa=('IN_ACESSIBILIDADE_RAMPAS','mean'),
)
```

```python
dt_gestor = pd.read_csv(
    "dados/Tabela_Gestor_Escolar_2025_v2.csv",
    sep=";",
    encoding="latin1"
)
```

```python
dt_matricula = pd.read_csv(
    "dados/Tabela_Matricula_2025_V2.csv",
    sep=";",
    encoding="latin1"
)
```

```python
mat_agg = dt_matricula.groupby(['NO_UF','TP_LOCALIZACAO','TP_DEPENDENCIA'], as_index=False).agg(
    total_mat=('QT_MAT_BAS','sum'),
    total_deficiencia=('QT_MAT_BAS_D','sum'),
    total_transporte=('QT_TRANSP_PUBLICO','sum')
)
mat_agg['pct_deficiencia'] = mat_agg['total_deficiencia'] / mat_agg['total_mat']
mat_agg['pct_transporte']  = mat_agg['total_transporte']  / mat_agg['total_mat']
```

```python
dt_turma = pd.read_csv(
    "dados/Tabela_Turma_2025_V2.csv",
    sep=';',
    encoding="latin1"
)
```

```python
tur_agg = dt_turma.groupby(['NO_UF','TP_LOCALIZACAO','TP_DEPENDENCIA'], as_index=False).agg(
    total_turmas=('QT_TUR_BAS','sum'),
    total_corrfluxo=('QT_TUR_FUND_AF_CORRFLUXO','sum'),
)
tur_agg['pct_turmas_corrfluxo'] = tur_agg['total_corrfluxo'] / tur_agg['total_turmas']
```
